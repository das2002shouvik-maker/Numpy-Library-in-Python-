## Numpy Library

```python
# Advantages of Numpy Library:
# 1. Code reusability.
# 2. Time efficient.
# 3. Used to handle complex codes.
```

```python
# Numpy stands for numerical python.
# Numpy is a library of python which is used to work with arrays. 
# An array is a data structure which comprises of a collection of similar values.
# In python we have list to do the same task, but utilizing an array using library enhances the task by an increase of 50x as compared to a list.
# Numpy was created in 2005 by Travis Oliphant.
```

```python
# Mean, Max, Min
marks = np.array([100,30,50,70,66])
marks.mean()
marks.max()
marks.min()
```

```python
# Flatten
a=np.array([[3,5,7,8],[6,3,8,9],[8,6,4,2]])
b=a.flatten()
print(b)
```

```python
# Generate Random Number

from numpy import random

a=random.randint(100)
print(a)
```

```python
# Generate a 1-D array containing 5 random integers from 0 to 100:

from numpy import random
a=random.randint(100, size=(5))
print(a)
```

```python
# Choice

from numpy import random
a=random.choice([5,6,7,9])
print(a)
```

```python
# Generate a 2-D array with 3 rows, each row containing 5 random integers from 0 to 100:
from numpy import random

f=random.randint(100, size=(3,5))
print(f)
```

```python
# Generate a 2-D array that consists of the values in the array parameter (3, 5, 7, and 9):

from numpy import random
x = random.choice([3, 5, 7, 9], size=(3, 5))
print(x)
```
